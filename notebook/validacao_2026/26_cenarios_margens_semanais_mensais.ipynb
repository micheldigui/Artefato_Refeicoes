## Leitura dos resultados

Em 2025, as margens reduziram o déficit potencial e aumentaram o excesso. O conservador levou a um aumento expressivo do excesso; ele constitui uma alternativa de maior proteção contra falta, sem justificar adoção automática. A prioridade 9:1 é uma hipótese, não uma avaliação de custos efetivamente pagos.

No diagnóstico de 2026 com histórico atualizado, o quantitativo institucional permaneceu com menor erro médio. O conservador mensal reduziu o déficit potencial frente à referência, com maior excesso e maior erro; o percentual observado de dias sem déficit ficou abaixo de 90%. A avaliação precisa apresentar essas diferenças, sem prometer superioridade em todas as métricas.

A coorte pareada tem 97, 108, 133 e 99 datas em 2023–2026. Ela exclui datas ausentes em um dos horizontes, dias sem cardápio, dias marcados como não letivos e datas sem quantitativo institucional. Não substitui as coortes anteriores de 390/397 datas. As antecedências continuam diferentes: previsão mensal emitida no dia 3 versus referência institucional semanal.

Excesso/déficit são diferenças potenciais diante do consumo observado; não comprovam desperdício físico ou refeições não atendidas. As emissões semanais e mensais devem ser avaliadas separadamente. As previsões puras de 05–09/10 diferem entre emissões por causa do corte de informação (02/10 semanal; 03/10 mensal), e não foram alteradas nesta etapa.


Na análise com defasagem, a coorte de 2023 passa a 79 datas; 2024/2025/2026 permanecem com 108/133/99. Em 2026, o conservador com defasagem alcançou 95,96% de dias sem déficit potencial, com excesso de 6.483 refeições e erro médio66,58. Isso não garante cobertura futura nem justifica adoção automática. Reexecutar esta etapa pontua as três emissões suplementares usando somente consumos aprovados na coleta existente, sem refazer previsões.
