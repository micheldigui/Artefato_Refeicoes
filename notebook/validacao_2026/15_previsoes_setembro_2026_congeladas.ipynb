# 15 — Previsões de setembro com consumo oculto

Gera setembro inteiro com alvos de treino até 2025 e histórico disponível até agosto. Cada lote usa origem na sexta-feira anterior e histórico estritamente anterior. É uma avaliação retrospectiva com alvos ocultos, produzida em outubro. O calendário/cardápio precisa ter sua disponibilidade histórica confirmada. As previsões de outubro são preservadas. A política primária é o plano puro; margens são cenários secundários. Não sobrescrever o congelamento.

Reexecução segura: se a emissão já existe, este notebook confere seu SHA-256 e exibe os dados preservados. O código de geração permanece disponível para a primeira emissão em uma nova janela.

```python
from pathlib import Path
import json,hashlib
import pandas as pd
_root_emissao=Path(r'F:\PRODUTO_MESTRADO')
_arquivo_emissao=_root_emissao/'resultados/validacao_setembro_2026/previsoes_setembro_2026_congeladas.csv'
_manifesto_emissao=_root_emissao/'resultados/validacao_setembro_2026/manifesto_setembro_2026.json'
PRESERVAR_EMISSAO_EXISTENTE=_arquivo_emissao.exists()
if PRESERVAR_EMISSAO_EXISTENTE:
    _manifesto=json.loads(_manifesto_emissao.read_text(encoding='utf-8'))
    _hash=hashlib.sha256(_arquivo_emissao.read_bytes()).hexdigest()
    assert _hash in json.dumps(_manifesto), 'Integridade da emissão congelada não confirmada.'
    print('Emissão existente validada por SHA-256 e preservada. Nenhuma nova emissão produzida.')
    display(pd.read_csv(_arquivo_emissao).head())
```

```python
if not PRESERVAR_EMISSAO_EXISTENTE:
    from pathlib import Path
    import runpy
    ROOT = Path(r'F:\PRODUTO_MESTRADO')
    runpy.run_path(str(ROOT / 'scripts' / 'prever_setembro_2026.py'), run_name='__main__')
```
